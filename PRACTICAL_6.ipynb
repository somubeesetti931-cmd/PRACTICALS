```python
import time

def matrix_chain_multiplication(p, n):
    # Create DP table
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # length is the chain length
    for length in range(2, n):
        for i in range(1, n - length + 1):
            j = i + length - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i - 1] * p[k] * p[j]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n - 1]


# User Input
n = int(input("Enter number of matrices: "))

p = []

print("Enter matrix dimensions:")

for i in range(n):
    if i == 0:
        rows = int(input(f"Enter rows of Matrix {i + 1}: "))
        p.append(rows)

    columns = int(input(f"Enter columns of Matrix {i + 1}: "))
    p.append(columns)


# Check if matrices can be multiplied
valid = True

for i in range(n - 1):
    if p[i + 1] != p[i + 1]:
        valid = False

# Calculate execution time
start_time = time.perf_counter()

minimum_cost = matrix_chain_multiplication(p, n)

end_time = time.perf_counter()

execution_time = end_time - start_time


# Output
print("\nMinimum number of scalar multiplications =", minimum_cost)
print("Time Complexity = O(n^3)")
print("Space Complexity = O(n^2)")
print("Execution Time = {:.9f} seconds".format(execution_time))
```
